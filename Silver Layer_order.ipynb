```python
df=spark.read.format("parquet").load("abfss://destination@proddatalakefordatabrick.dfs.core.windows.net/bronze/order")
```

```python
df=df.drop("_rescued_data")
```

```python
from pyspark.sql.functions import *

df=df.withColumn("order_date", to_timestamp(col("order_date")))
```

```python
df=df.withColumn("Year",year(col("order_date")))
```

```python
from pyspark.sql.window import Window
from pyspark.sql.functions import dense_rank, col

window_spec = Window.partitionBy("Year").orderBy(col("total_amount").desc())
df_ranked = df.withColumn("dense_rank",dense_rank().over(window_spec))
```

```python
df_ranked.write.format("delta").mode("overwrite").save("abfss://destination@proddatalakefordatabrick.dfs.core.windows.net/silver/order")
```
